# Fermat's manual prime testing

Peter Shui's “Fermat's Method of Factorization” (*Mathematic Gazette, 2015) works out a plausible way in which Fermat could have factored 100895598169 “in a day” as Fermat wrote to Mersenne.
But it does not address how Fermat determined whether numbers were prime.
For that, Fermat would have used what is no called Fermat's *Little* Theorem (FLT),
which I will state as 

Fermat's little theorem

If *p* is prime then for all $a$ such that $\gcd(a, p) = 1$

\begin{equation}
    a^{p-1} \equiv 1 \pmod p \tag{FLT}
\end{equation}

Note that this can hold for composite numbers as well as primes, but it must hold of primes.



In particular Fermat would have checked whether 

\begin{equation}
    a^{100895598169 - 1} \not\equiv 1 \pmod{100895598169}
\end{equation}

for some convenient values of *a*.

The most convient value of *a* to try will be 2.
And we will use $M$ (for Mersenne's challnage) as 100895598169

```python
a = 2
M = 100895598169
```

And we can see that uses $a=2$ makes 2 a witness to the compositeness of $M$.

```python
pow(a, M - 1, M)
```

How might Fermat have computed that? This determination would be a small part compared to the actual work of factoring *M*.


I will assume that Fermat already had a table of powers of 2, and so would be able to quickly find the smallest power of 2 greater than M. I will use a computer for that.

```python
from math import log2

b = int(log2(M)) + 1
print(f'2^{b} = {1 << b}')
```

With that we can restructure the problem as testing whether

\begin{equation}
    2^{37} 2^{M - 38} \not\equiv \pmod M
\end{equation}

So lets work on $2^{37} \bmod M$, which we will call $B$.
If we have constucted $b$ correctly then $B$ will simply be

\begin{equation}
\begin{aligned}
B &= 2^b - M\\
  &= 2^{37} - M  \\
  &= 137438953472 - M\\
  &= 137438953472 - 100895598169\\
\end{aligned}
\end{equation}

```python
B = 137438953472 - 100895598169
assert B == pow(2, 37, M)
print(f'B = {B}')
```

That is an unfortunately large number. Let's see how big it is compared to *M*.

```python
B/M
```

Ok. So it is a little more than 1/3 the size of *M*. That means that $B_3 = 3B \mod M$ should be a smaller and more managable number, and should be computable with a few few multiplications and subtrations. That is

\begin{equation}
    B_3 = 3B \mod M = 3B - M
\end{equation}

```python
B_3 = (3 * B) % M
B_3
```

```python
B_3/M
```

And just to check my logic

```python
B_3 == 3 * B - M
```

Now some playing with properties of exponents and modular equivalence.

$$
B \equiv 2^b - M \equiv 2^b \pmod M
$$

```python
B == pow(2, b, M)
```
